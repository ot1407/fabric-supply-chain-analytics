```python
file_path = "Files/bronze/DataCoSupplyChainDataset.csv"

df_bronze = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(file_path)
)
```

```python
columns_to_drop = [
    "Customer Password",
    "Customer Email",
    "Product Description"
]

df_silver = df_bronze.drop(*columns_to_drop)
```

```python
print("Bronze Columns :", len(df_bronze.columns))
print("Silver Columns :", len(df_silver.columns))
```

```python
for col in df_silver.columns:
    new_col = (
        col.lower()
           .replace(" ", "_")
           .replace("(", "")
           .replace(")", "")
    )

    df_silver = df_silver.withColumnRenamed(col, new_col)
```

```python
df_silver.columns
```

```python
print("Silver Columns:", len(df_silver.columns))
print("Silver Rows:", df_silver.count())
```

```python
df_silver.write.mode("overwrite").saveAsTable("silver_supply_chain")
```
