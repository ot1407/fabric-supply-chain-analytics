```python
df_silver = spark.table("silver_supply_chain")
```

```python
display(df_silver)
```

```python
df_silver.count()
```

```python
from pyspark.sql.functions import count, avg, sum, round

gold_delivery_performance = (
    df_silver
    .groupBy("delivery_status")
    .agg(
        count("*").alias("order_count"),
        round(avg("days_for_shipping_real"), 2).alias("avg_actual_shipping_days"),
        round(avg("days_for_shipment_scheduled"), 2).alias("avg_scheduled_shipping_days"),
        round(sum("sales_per_customer"), 2).alias("total_sales")
    )
    .orderBy("delivery_status")
)

display(gold_delivery_performance)
```

```python
gold_delivery_performance.write.mode("overwrite").saveAsTable(
    "gold_delivery_performance"
)
```
